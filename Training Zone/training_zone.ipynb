```python
# Importing useful dependencies
import boto3
```

```python
# Setup S3 client for MinIO (MinIO implements Amazon S3 API)
s3 = boto3.client(
    "s3",
    #endpoint_url="http://127.0.0.1:9000", # MinIO API endpoint
    endpoint_url="https://distractively-polygalaceous-lindsey.ngrok-free.dev", # MinIO API endpoint (ngrok)
    aws_access_key_id="minioadmin", # User name
    aws_secret_access_key="minioadmin", # Password
)
```

```python
# List existing buckets
buckets = [b["Name"] for b in s3.list_buckets()["Buckets"]]

# Function that given a name, creates a bucket
def createBucket(name, list_buckets):
    if name in list_buckets:
        print(f"Bucket '{name}' already exists!")
    else:
        s3.create_bucket(Bucket=name)
        print(f"Created bucket: {name}")

# Create a bucket named training zone
createBucket("training-zone", buckets)
```

```python
# Create three sub-buckets inside training_zone.
# Each of these sub-buckets will be used to store model weights
s3.put_object(Bucket="training-zone", Key="FP16-INT4/")
s3.put_object(Bucket="training-zone", Key="LoRA/")
s3.put_object(Bucket="training-zone", Key="FP16-INT4-LoRA   /")
```
